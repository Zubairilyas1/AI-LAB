```python
v = [
    {"id": "A01", "zone": "A", "status": "occupied", "rate": 100},
    {"id": "A02", "zone": "A", "status": "available", "rate": 100},
    {"id": "B01", "zone": "B", "status": "occupied", "rate": 80},
    {"id": "B02", "zone": "B", "status": "available", "rate": 80},
    {"id": "C01", "zone": "C", "status": "available", "rate": 120},
    {"id": "C02", "zone": "C", "status": "occupied", "rate": 120}
]

print("Available SPACE: ")
a = 0   # Count of available spaces
b = 0   # Count of occupied spaces
cheapest_rate = v[0]["rate"]
cheapest_id = v[0]["id"]

for space in v:
    if space["status"] != "occupied":
        a += 1
        if cheapest_rate > space["rate"]:
            cheapest_rate = space["rate"]
            cheapest_id = space["id"]
        print("spot: ", space["id"])
    else:
        b += 1

print("\nAvailable Parking:", a)
print("Occupied:", b)
print("Cheapest Available spot:", cheapest_id, "at", cheapest_rate)

# ---------- FUNCTION (Now working correctly) ----------
def cheap(rate, hours):
    total_cost = rate * hours
    return total_cost   # Return the result, don't print inside

# ---------- USER INPUT ----------
user_hours = 3  # Ask ONCE outside
user_rate = cheapest_rate                            # Use the cheapest rate

result = cheap(user_rate, user_hours)   # Pass both values
print("Total price for", user_hours, "hours at", user_rate, "PKR/hour:", result)
```
