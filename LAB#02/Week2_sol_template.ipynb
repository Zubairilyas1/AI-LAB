### TASK 3: Missing Data Investigation and Imputation

```python
# Step 1: Load Dataset
import seaborn as sns
import pandas as pd

penguins_df = sns.load_dataset("penguins")
```

```python
# Step 2: Inspect the dataset
print("Shape:", penguins_df.shape)
print("Missing counts:\n", penguins_df.isnull().sum())
```

```python
# Step 4: Handle numerical missing values
num_cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
for col in num_cols:
    penguins_df[col] = penguins_df[col].fillna(penguins_df[col].mean())
```

```python
# Step 5: Handle categorical missing values

cat_cols = ["species", "island", "sex"]
for col in cat_cols:
    penguins_df[col] = penguins_df[col].fillna(penguins_df[col].mode()[0])
```

```python
# Step 6: Verify that there are no more missing values
print("Remaining missing values:",penguins_df.isnull().sum())

```
