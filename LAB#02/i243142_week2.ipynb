```python
# Step 1: Load Dataset
import seaborn as sns
import pandas as pd

penguins_df = sns.load_dataset("penguins")
```

```python
num_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g",
]
```

```python
# Step 4: Handle numerical missing values
for col in num_cols:
    median_val = penguins_df[col].median()
    penguins_df[col] = penguins_df[col].fillna(median_val)
```

```python
# Step 5: Handle categorical missing values
mode_sex = penguins_df["sex"].mode()[0]
penguins_df["sex"] = penguins_df["sex"].fillna(mode_sex)
```

```python
# Step 6: Verify that there are no more missing values
print("Remaining missing values:")
print(penguins_df.isnull().sum())
```

# PHASE 4: INVESTIGATE OUTLIERS

### TASK 7: Statistical Outlier Identification Using IQR

```python
def iqr_analysis(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return Q1, Q3, IQR, lower_bound, upper_bound, len(outliers)
```

```python
# SalePrice
saleprice_bounds = iqr_analysis(
    house_data,
    "SalePrice"
)
```

```python
# GrLivArea
GrLivArea = iqr_analysis(house_data, "GrLivArea")
```

```python
# GarageArea
GarageArea = iqr_analysis(house_data, "GarageArea")
```

```python
# Box Plots
features = ["SalePrice", "GrLivArea", "GarageArea"]
results = {}
for feat in features:
    q1, q3, iqr, lb, ub, n_out = iqr_analysis(house_data, feat)
    results[feat] = {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lb,
        "Upper Bound": ub,
        "Outlier Count": n_out,
    }

```

```python
iqr_df = pd.DataFrame(results).T
display(iqr_df)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, feat in enumerate(features):
    sns.boxplot(y=house_data[feat], ax=axes[i], color="lightblue")
    axes[i].set_title(f"Box Plot: {feat}")
plt.tight_layout()
plt.show()
```

### TASK 8: Outlier Mitigation Through IQR-Based Filtering

```python
# using task 7 calculated bounds
sp_lb, sp_ub = (
    results["SalePrice"]["Lower Bound"],
    results["SalePrice"]["Upper Bound"],
)
gr_lb, gr_ub = (
    results["GrLivArea"]["Lower Bound"],
    results["GrLivArea"]["Upper Bound"],
)

# Apply IQR Filtering to build clean_house_df
clean_house_df = house_data[
    (house_data["SalePrice"] >= sp_lb)
    & (house_data["SalePrice"] <= sp_ub)
    & (house_data["GrLivArea"] >= gr_lb)
    & (house_data["GrLivArea"] <= gr_ub)
].copy()

# Document Shape Changes
orig_rows = house_data.shape[0]
clean_rows = clean_house_df.shape[0]
removed_rows = orig_rows - clean_rows
pct_removed = (removed_rows / orig_rows) * 100

print(f"Original Dataset Shape: {house_data.shape}")
print(f"Cleaned Dataset Shape:  {clean_house_df.shape}")
print(
    f"Observations Removed:   {removed_rows} ({pct_removed:.2f}% of total data)"
)
```
