```python
class ModelBasedPatientAgent:
    """
    Model-Based Agent that maintains internal memory of past vital sign readings
    to detect abnormal deteriorating trends in partially observable environments.
    """
    def __init__(self):
        self.history = []  # Internal state memory

    def update_internal_state(self, current_percept):
        self.history.append(current_percept)

    def decide_action(self, current_percept):
        self.update_internal_state(current_percept)
        
        hr = current_percept.get("heart_rate", 80)
        bp_sys = current_percept.get("bp_systolic", 120)
        spo2 = current_percept.get("spo2", 98)

        # 1. Critical Threshold Check
        if spo2 < 85 or hr > 130 or bp_sys < 90:
            return "Call Doctor (Critical Trend Detected)"
        
        # 2. Historical Trend Analysis (Internal Memory Evaluation)
        if len(self.history) > 1:
            prev_reading = self.history[-2]
            prev_hr = prev_reading.get("heart_rate", 80)
            prev_spo2 = prev_reading.get("spo2", 98)
            
            # Detect persistent worsening trend: Increasing HR & Decreasing SpO2
            if hr > prev_hr and spo2 < prev_spo2:
                return "Alert Nurse (Abnormal Deteriorating Trend)"

        # 3. Mild Abnormality Check
        if hr > 100 or bp_sys > 140 or spo2 < 95:
            return "Alert Nurse (Slightly Abnormal)"
            
        return "Continue Monitoring (Stable)"

# Simulation Execution
patient_agent = ModelBasedPatientAgent()
patient_vitals_over_time = [
    {"time": "t-2", "heart_rate": 98,  "bp_systolic": 130, "spo2": 95},
    {"time": "t-1", "heart_rate": 105, "bp_systolic": 140, "spo2": 91},
    {"time": "t0",  "heart_rate": 112, "bp_systolic": 100, "spo2": 82}
]

print("--- Task 2: Patient Monitoring Simulation ---")
for vitals in patient_vitals_over_time:
    action = patient_agent.decide_action(vitals)
    print(f"Time {vitals['time']} | Sensor Reading: {vitals} --> Action: {action}")
```

```python
class GoalBasedWarehouseAgent:
    """
    Goal-Based Agent that formulates navigation steps to reach a predefined destination.
    """
    def __init__(self, goal_location=(3, 3)):
        self.goal = goal_location  # Predefined operational goal (Loading Dock)

    def is_goal_achieved(self, current_location):
        return current_location == self.goal

    def choose_action(self, current_location):
        if self.is_goal_achieved(current_location):
            return "STOP (Goal Achieved)"
        
        curr_x, curr_y = current_location
        goal_x, goal_y = self.goal
        
        if curr_x < goal_x: return "Move Right"
        elif curr_x > goal_x: return "Move Left"
        elif curr_y < goal_y: return "Move Up"
        elif curr_y > goal_y: return "Move Down"

# Simulation Execution
warehouse_agent = GoalBasedWarehouseAgent(goal_location=(3, 3))
initial_positions = [(0, 0), (1, 3), (3, 3)]

print("--- Task 3: Autonomous Warehouse Navigation Simulation ---")
for start in initial_positions:
    current_pos = start
    print(f"\nSimulating path starting from initial location: {start}")
    step_count = 0
    while not warehouse_agent.is_goal_achieved(current_pos) and step_count < 10:
        action = warehouse_agent.choose_action(current_pos)
        print(f"  Current Position: {current_pos} -> Action Selected: {action}")
        if action == "Move Right": current_pos = (current_pos[0] + 1, current_pos[1])
        elif action == "Move Left": current_pos = (current_pos[0] - 1, current_pos[1])
        elif action == "Move Up": current_pos = (current_pos[0], current_pos[1] + 1)
        elif action == "Move Down": current_pos = (current_pos[0], current_pos[1] - 1)
        step_count += 1
    print(f"  Final Location: {current_pos} | Goal Reached: {warehouse_agent.is_goal_achieved(current_pos)}")
```

```python
class UtilityBasedCloudAgent:
    """
    Utility-Based Agent performing multi-criteria optimization.
    """
    def __init__(self, w_perf=0.5, w_cost=0.3, w_lat=0.2):
        self.w_perf = w_perf
        self.w_cost = w_cost
        self.w_lat = w_lat

    def compute_utility(self, server_specs):
        # Utility = (w_perf * Performance) - (w_cost * Cost) - (w_lat * Latency)
        return round((self.w_perf * server_specs['cpu_perf']) - \
                     (self.w_cost * server_specs['cost']) - \
                     (self.w_lat * server_specs['latency']), 2)

    def select_best_server(self, servers):
        best_server = None
        highest_utility = float('-inf')
        
        for name, specs in servers.items():
            u_score = self.compute_utility(specs)
            print(f"  {name} specs={specs} -> Utility U = {u_score}")
            if u_score > highest_utility:
                highest_utility = u_score
                best_server = name
                
        return best_server, highest_utility

# Simulation Execution
available_servers = {
    'Server A': {'cost': 12, 'cpu_perf': 90, 'latency': 40},
    'Server B': {'cost': 8,  'cpu_perf': 85, 'latency': 20},
    'Server C': {'cost': 15, 'cpu_perf': 98, 'latency': 35}
}

print("--- Task 4: Utility-Based Server Selection Simulation ---")
cloud_agent = UtilityBasedCloudAgent(w_perf=0.5, w_cost=0.3, w_lat=0.2)
selected_server, score = cloud_agent.select_best_server(available_servers)
print(f"\nOptimal Decision: Choose {selected_server} with maximum utility score ({score})")
```

```python
class SimpleReflexIrrigationAgent:
    """
    Simple Reflex Agent operating purely on instantaneous condition-action rule lookup.
    """
    def act(self, moisture_level):
        if moisture_level < 30:
            return "Pump ON (Dry Soil)"
        elif 30 <= moisture_level <= 60:
            return "Pump OFF (Moist Soil - Optimal)"
        else:
            return "Pump OFF (Wet Soil)"

# Simulation Execution
irrigation_agent = SimpleReflexIrrigationAgent()
test_moisture_readings = [20, 55, 82]

print("--- Task 5: Simple Reflex Irrigation Control Simulation ---")
for reading in test_moisture_readings:
    action = irrigation_agent.act(reading)
    print(f"Percept (Soil Moisture): {reading}% -> Actuator Action: {action}")
```

```python
class DeterministicManufacturingEnvironment:
    """
    Deterministic environment model where actions lead to predictable physical state transitions.
    """
    def step(self, current_state, action):
        transitions = {
            ("Initial", "Pick Bottle"): "Bottle in Gripper",
            ("Bottle in Gripper", "Place Bottle"): "Bottle in Box"
        }
        return transitions.get((current_state, action), current_state)

# Simulation Execution
manufacturing_env = DeterministicManufacturingEnvironment()
action_sequence = ["Pick Bottle", "Place Bottle"]

print("--- Task 6: Deterministic Manufacturing Environment Simulation ---")
for run in range(1, 4):
    state = "Initial"
    print(f"\nExecution Run {run}:")
    for action in action_sequence:
        state = manufacturing_env.step(state, action)
        print(f"  Command: '{action}' -> Next State: '{state}'")
    print(f"  Run {run} Final Outcome: {state} (100% Deterministic Consistency)")
```

```python
import random

class StochasticDroneEnvironment:
    """
    Stochastic environment model where identical agent commands produce probabilistic results.
    """
    def step(self, current_state, action):
        if action == "Fly to Destination":
            outcomes = [
                ("Delivered On Time (Clear Weather)", 0.60),
                ("Delivered with Delay (Strong Wind)", 0.25),
                ("Rerouted / Delay (GPS Signal Loss)", 0.15)
            ]
            
            r = random.random()
            cumulative_prob = 0.0
            for outcome_text, prob in outcomes:
                cumulative_prob += prob
                if r <= cumulative_prob:
                    return outcome_text
        return current_state

# Simulation Execution
drone_env = StochasticDroneEnvironment()
command = "Fly to Destination"

print("--- Task 7: Stochastic Drone Delivery Simulation ---")
print(f"Command Executed: '{command}' across 5 independent delivery flights:")
random.seed(42)  # Set seed for reproducible stochastic demonstration
for flight_id in range(1, 6):
    result = drone_env.step("Base Station", command)
    print(f"  Flight {flight_id} Result -> {result}")
```
